##### BIVARIATE ANALYSIS ( Is gender has any impact on student's performance ? )

```python
gender_group = df.groupby('gender')['average'].mean()
gender_group
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# --------------------------------------------------
# 1. Agar 'average' column already nahi hai,
#    to average calculate karo
# --------------------------------------------------

if 'average' not in df.columns:
    df['average'] = df[['math_score', 'reading score', 'writing score']].mean(axis=1)

# --------------------------------------------------
# 2. Gender-wise average
# --------------------------------------------------

gender_group = df.groupby('gender')[['average', 'math_score']].mean()

print(gender_group)

# --------------------------------------------------
# 3. Values nikalna
# --------------------------------------------------

female_scores = [
    gender_group.loc['female', 'average'],
    gender_group.loc['female', 'math_score']
]

male_scores = [
    gender_group.loc['male', 'average'],
    gender_group.loc['male', 'math_score']
]

# --------------------------------------------------
# 4. X-axis labels
# --------------------------------------------------

X = ['Total Average', 'Math Average']
X_axis = np.arange(len(X))

# --------------------------------------------------
# 5. Bar width
# --------------------------------------------------

width = 0.35

# --------------------------------------------------
# 6. Plot
# --------------------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    X_axis - width/2,
    female_scores,
    width,
    label='Female'
)

plt.bar(
    X_axis + width/2,
    male_scores,
    width,
    label='Male'
)

# --------------------------------------------------
# 7. Labels and formatting
# --------------------------------------------------

plt.xticks(X_axis, X)
plt.xlabel('Score Type')
plt.ylabel('Average Score')
plt.title('Female vs Male Average Scores')

plt.legend()

plt.tight_layout()
plt.show()
```

##### Insights
- On an average females have a better overall score than men.
- whereas males have scored higher in Maths.

##### BIVARIATE ANALYSIS ( Is Race/Ehnicity has any impact on student's performance ? )

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group by race/ethnicity
Group_data2 = df.groupby('race_ethnicity')

# Create 3 plots
f, ax = plt.subplots(1, 3, figsize=(20, 8))

# -------------------------
# 1. Math Score
# -------------------------

math_mean = Group_data2['math_score'].mean()

sns.barplot(
    x=math_mean.index,
    y=math_mean.values,
    hue=math_mean.index,
    palette='mako',
    legend=False,
    ax=ax[0]
)

ax[0].set_title('Math Score', color='#005ce6', size=20)
ax[0].set_xlabel('Race/Ethnicity')
ax[0].set_ylabel('Average Score')

for container in ax[0].containers:
    ax[0].bar_label(
        container,
        color='black',
        size=12,
        fmt='%.1f'
    )


# -------------------------
# 2. Reading Score
# -------------------------

reading_mean = Group_data2['reading_score'].mean()

sns.barplot(
    x=reading_mean.index,
    y=reading_mean.values,
    hue=reading_mean.index,
    palette='flare',
    legend=False,
    ax=ax[1]
)

ax[1].set_title('Reading Score', color='#005ce6', size=20)
ax[1].set_xlabel('Race/Ethnicity')
ax[1].set_ylabel('Average Score')

for container in ax[1].containers:
    ax[1].bar_label(
        container,
        color='black',
        size=12,
        fmt='%.1f'
    )


# -------------------------
# 3. Writing Score
# -------------------------

writing_mean = Group_data2['writing_score'].mean()

sns.barplot(
    x=writing_mean.index,
    y=writing_mean.values,
    hue=writing_mean.index,
    palette='coolwarm',
    legend=False,
    ax=ax[2]
)

ax[2].set_title('Writing Score', color='#005ce6', size=20)
ax[2].set_xlabel('Race/Ethnicity')
ax[2].set_ylabel('Average Score')

for container in ax[2].containers:
    ax[2].bar_label(
        container,
        color='black',
        size=12,
        fmt='%.1f'
    )


# Layout
plt.tight_layout()
plt.show()
```

##### Insights
- Group E students have scored the highest marks.
- Group A students have scored the lowest marks.
- Students from a lower Socioeconomic status have a lower avg in all course subjects
